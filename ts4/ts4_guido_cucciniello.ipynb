# Tiempo de retardo
Supongamos que tenemos un sistema el cual desplaza en el tiempo una señal de entrada de la siguiente forma
<figure style="text-align: center;">     
    <img src="images/retardo.png" alt="Descripción de la imagen" style="width: 70%;">
    <figcaption>Figura 1: A modo ilustrativo, se muestra el funcionamiento principal de un Sistema de Retardo.</figcaption>
</figure>

Por Fourier sabemos que podemos descomponer cualquier señal en una suma infinita de señales sinusoidales, por esto, es de sumo interés empezar el análisis con una señal de entrada senoidal pura. Digamos que la señal de entrada a un sistema de retardo es $$v_1(t) = A\,\sin(\omega\,t + \phi)$$ cuya señal de salida es $$v_2(t) = A\,\sin(\omega\,(t - D_f) + \phi)$$

La entrada y la salida solo difieren en un ángulo de fase $$\theta(\omega) = -\omega\,D_f$$

De forma fasorial se puede expresar como $$\frac{V_2}{V_1} = 1 \,\angle -\omega\,D_f$$ $$T(j\,\omega) = \frac{V_2(j\omega)}{V_1(j\omega)} = e^{-j\omega\,D_f} = e^{-j\,\omega\,\frac{1}{\frac{1}{D_f}}}$$

Si graficamos el comportamiento de la fase del sistema de retardo, esta es constante y no existe distorsión de fase a ninguna frecuencia que se considere.
<figure style="text-align: center;">     
    <img src="images/RETARDO_IDEAL.png" alt="Descripción de la imagen" style="width: 70%;">
    <figcaption>Figura 2: Fase de un sistema de Retardo Ideal</figcaption>
</figure>

Se toma como norma de frecuencia $\Omega_\omega = \omega_c = \frac{1}{D_f} \quad \Rightarrow \quad T(j\omega') = e^{-j\frac{\omega}{\omega_c}} = e^{-j\omega'}$ 

Generalizando en el plano complejo $s' = j\omega'$, tenemos $$T(s') = e^{-s'}$$
Entonces, volviendo al dominio de la frecuencia normalizada $T(j\omega') = T(s') \Big|_{s' = j\omega'} \quad \Rightarrow \quad T(j\omega') = R(\omega') + jX(\omega')$ cuya fase es $$\theta(\omega') = \arctan\left(\frac{X(\omega')}{R(\omega')}\right)$$
Igualando esta expresión con la fase de un sistema de retardo constante se llega a que $$-\omega'\,D_f = \arctan\left(\frac{X(\omega')}{R(\omega')}\right)$$ y despejando el retardo de fase queda $$D_f(\omega')= -\frac{1}{\omega'}\,\arctan\left(\frac{X(\omega')}{R(\omega')}\right)$$
Esta expresión nos indica que el retardo es una función transcendental, es decir, no satisface una ecuación polinómica cuyos coeficientes son función de la frecuencia. A la hora de diseñar filtros, estos responden con transferencias racionales (cocientes de polinomios en $s'$), y sabiendo que $D_f(\omega')$ está gobernado por una función transcendental, es imposible que un sistema de orden finito presente un retardo de fase constante para todo el espectro de frecuencias. Por lo tanto el diseño se limitará a una aproximación óptima en un ancho de banda de interés.

La expresión de $D_f(\omega')$ no es práctica para diseñar filtros reales. Si la señal que estamos tratando presenta varias componentes de frecuencia, cada una de ellas se retrasaría individualmente un tiempo distinto, esto causa distorsión de fase. Por este motivo, se define el retardo de grupo como la tasa de cambio de la fase respecto de la frecuencia $$D_g(\omega') = -\frac{d\,\theta(\omega')}{d\,\omega'}$$
Lo que interesa es que dicha tasa de cambio sea constante, y no que cada retardo individual de cada componente de frecuencia sea el mismo.

Recordando que $\frac{d}{du}\,\arctan(u) = \frac{1}{1 + u^{2}}\,\frac{du}{d\omega'}$ siendo $u(\omega') = \frac{X(\omega')}{R(\omega')}$ entonces $$D_g(\omega') = \frac{X\,\frac{dR}{d\omega'} - R\,\frac{dX}{d\omega'}}{R^{2} + X^{2}}$$ Queda de manera explícita que el retardo de grupo jamás será constante, puesto que hay variaciones con respecto a la frecuencia. Entonces si la fase nunca resultará lineal de modo tal que el retardo de grupo sea constante, habrá que aproximarla a lineal de alguna forma u otra.

# Aproximación Bessel-Thomson
Podemos aproximar la función transferencia $T(s') = e^{-s'}$ de manera racional utilizando el método de Storch, el cual permite generar los polinomios de Bessel de manera recursiva. 

La función transferencia que queremos aproximar es $$T(s') = \frac{1}{e^{s'}}$$ Aprovechando las funciones hiperbólicas: 
\begin{align}
    \cosh(s') &= \frac{e^{s'} + e^{-s'}}{2} \\\\
    \sinh(s') &= \frac{e^{s'} - e^{-s'}}{2}
\end{align}

Sumando dichas funciones obtenemos que $$e^{s'} = \cosh(s') + \sinh(s')$$ entonces 
\begin{align}
     T(s') &= \frac{1}{\cosh(s') + \sinh(s')}\\\\
     T(s') &= \frac{\frac{1}{\sinh(s')}}{1 + \coth(s')}
\end{align}

La expansión en serie de estas funciones alrededor de $\omega' = 0$ son 
\begin{align}
     \cosh(s') &= 1 + \frac{s'^{2}}{2!} + \frac{s'^{4}}{4!} + \frac{s'^{6}}{6!} + \cdot \cdot \cdot \\\\
     \sinh(s') &= s' + \frac{s'^{3}}{3!} + \frac{s'^{5}}{5!} + \frac{s'^{7}}{7!} + \cdot \cdot \cdot
\end{align}
Se obtiene $$\coth(s') = \frac{1}{s'} + 
\cfrac{1}{
    \cfrac{3}{s'} + 
    \cfrac{1}{
        \cfrac{5}{s'} + 
        \cfrac{1}{
            \cfrac{7}{s'} + 
            \cfrac{1}{
                \cfrac{9}{s'} + 
                \cfrac{1}{
                    \cfrac{11}{s'} + 
                    \cfrac{1}{
                        \cfrac{13}{s'} + \dots
                    }
                }
            }
        }
    }
}$$

Al truncar esta fracción en un nivel $n$, se obtiene una aproximación racional con polinomios en $s'$ que aproxima el retardo ideal.

## Transferencias de Bessel
Obtener las transferencias de Bessel para $n = 2$, $n = 3$ y $n = 4$

<u> n = 2: </u>
$$\coth(s')=\frac{1}{s'} + \frac{1}{\frac{3}{s'}} = \frac{1}{s'} + \frac{s'}{3} = \frac{s'^{2} + 3}{s3}=\frac{\cosh(s')}{\sinh(s')}$$
Entonces construyendo la transferencia $$T_2(s') = \frac{\frac{1}{s'3}}{1 + \frac{s'^{2} + 3}{s'3}} = \frac{1}{s'^{2} + s'3 + 3}$$ y para garantizar que $|T_2(j\omega')| \Big|_{\omega' = 0} = 1$, necesariamente $$T_2(s') =\frac{3}{s'^{2} + s'3 + 3}$$

<u> n = 3: </u>
$$\coth(s')=\frac{1}{s'} + \frac{1}{\frac{3}{s'} + \frac{1}{\frac{5}{s'}}} = \frac{s'^{2}\,6 + 15}{s'\,(s'^{2} + 15)}=\frac{\cosh(s')}{\sinh(s')}$$
Entonces construyendo la transferencia $$T_3(s') = \frac{\frac{1}{s'\,(s'^{2} + 15)}}{1 + \frac{s'^{2}\,6 + 15}{s'\,(s'^{2} + 15)}} = \frac{1}{s'^{3} + s'^{2}\,6 + s'\,15 + 15}$$ y para garantizar que $|T_3(j\omega')| \Big|_{\omega' = 0} = 1$, necesariamente $$T_3(s') = \frac{15}{s'^{3} + s'^{2}\,6 + s'\,15 + 15}$$

<u> n = 4: </u>
$$\coth(s')=\frac{1}{s'} + \frac{1}{\frac{3}{s'} + \frac{1}{\frac{5}{s'} + \frac{1}{\frac{7}{s'}}}} = \frac{s'^{4} + s'^{2}\,45 + 105}{s'^{3}\,10 + s'\,105}=\frac{\cosh(s')}{\sinh(s')}$$
Entonces construyendo la transferencia $$T_4(s') = \frac{\frac{1}{s'^{3}\,10 + s'\,105}}{1 + \frac{s'^{4} + s'^{2}\,45 + 105}{s'^{3}\,10 + s'\,105}} = \frac{1}{s'^{4} + s'^{3}\,10 + s'^{2}\,45 + s'\,105 + 105}$$ y para garantizar que $|T_4(j\omega')| \Big|_{\omega' = 0} = 1$, necesariamente $$T_4(s') = \frac{105}{s'^{4} + s'^{3}\,10 + s'^{2}\,45 + s'\,105 + 105}$$

De formá genérica y recursiva $$B_n(s’)=\frac{B_n(0)}{B_n(s’)}$$ donde los polinomios de Bessel son $$B_n(s’) = (2n-1)\,B_{n-1}(s’) + s’^{2}\,B_{n-2}(s’)$$

### Retardo de Grupo $D_g(\omega')$
Para el cálculo del retardo de grupo hay que pasarnos al dominio de la frecuencia normalizada $\omega'$, para eso $$T_3(j\,\omega') = T_3(s') \Big|_{s' = j\,\omega'} = \frac{15}{(j\,\omega')^{3} + (j\,\omega')^{2}\,6 + (j\,\omega')\,15 + 15} = \frac{15}{15 - 6\omega'^{2} + j\,(15\omega' - \omega'^{3})}$$
La fase de esta transferencia es $$\theta_3(\omega')= -\arctan\left(\frac{15\omega' - \omega'^{3}}{15 - 6\omega'^{2}}\right)$$ y obteniendo el retardo de grupo
\begin{align}
    D_{g_3}(\omega') &= -\frac{d\,\theta_3(\omega')}{d\,\omega'} \\\\
    D_{g_3}(\omega') &= \frac{6\omega'^{4} + 45\omega'^{2} + 225}{\omega'^{6} + 6\omega'^{4} + 45\omega'^{2} + 225}
\end{align}
Evaluando al retardo de grupo en $\omega' = 0$ y $\omega' = 2.5$
\begin{align}
D_{g_3}(\omega') \Big|_{\omega' = 0} &= 1\,\text{seg} \\\\
D_{g_3}(\omega') \Big|_{\omega' = 2.5} &= 0.7521\,\text{seg}
\end{align}
Esto implica que $$D_{g_3}(2.5) = 0.7521\,D_{g_3}(0)$$
el retardo de grupo a una frecuencia normalizada de $2.5$ es el $75.21\%$ del valor máximo. En términos de error porcentual
$$\text{error} = (1 - 0.7521)\,100 \% = 24,79 \%$$

### Normalización
Como estamos trabajando en el dominio normalizado, la norma de frecuencia general para todo el circuito es la pulsación angular de paso $\omega_p$ especificada en la plantilla, es decir $\Omega_\omega = \omega_p\,\left[\frac{\text{rad}}{\text{s}}\right]$. Una vez aclarado esto
#### Normalización Sección de Orden 1

La pulsación angular de corte normalizada del filtro es $$\omega_{0_1}' = 2.32219$$
Sabiendo que $\omega_{0_1} = \frac{1}{R\,C} = \frac{1}{R'\,\Omega_Z\,\frac{C'}{\Omega_Z\,\Omega_\omega}} = \frac{\Omega_\omega}{R'\,C'} \quad \Rightarrow \quad \omega_{0_1}' = \frac{\omega_{0_1}}{\Omega_\omega} = \frac{1}{R'\,C'} = 2.32219$

Propongo que $C' = 1$ entonces $R' = 0.4306$

#### Normalización Sección de Orden 2
Propongo como norma de impedancia $\Omega_Z = R_1\,[\Omega]$
La pulsación angular de corte normalizada del filtro es $$\omega_{0_2}' = \sqrt{6.45944} = 2.5414$$
Sabiendo que $\omega_{0_2} = \frac{1}{\sqrt{R1R2C1C2}} = \frac{1}{\sqrt{R1'\,\Omega_Z\,R2'\,\Omega_Z\frac{C1'}{\Omega_Z\,\Omega_\omega}\,\frac{C2'}{\Omega_Z\,\Omega_\omega}}} = \frac{\Omega_\omega}{\sqrt{R1'R2'C1'C2}} \quad \Rightarrow \quad \omega_{0_2}' = \frac{\omega_{0_2}}{\Omega_\omega} = \frac{1}{\sqrt{R1'R2'C1'C2}} = 2.5414$ 


Propongo que $R_2' = 1$. Simulatáneamente , debemos hallar el valor de $C_1'$ y $C_2'$ que garantice la frecuencia $\omega_{0_2}'$ y el $Q = \frac{\sqrt{\frac{R1'R2'C2'}{C1'}}}{R1' + R2'}= 0.6910$ del circuito, obtenido a partir de la transferencia.

### Desnormalización
Se propone desnormalizar el filtro para tener un retardo de grupo de $200\,\mu\text{seg}$ en el origen $\omega = 0$
Hay que adoptar una nueva norma de frecuencia, para eso partamos de $$D_g(\omega) = \frac{D_g(\omega')}{\Omega_\omega}$$ 
La nueva norma de frecuencia será $$\Omega_\omega = \frac{D_g(\omega'= 0)}{D_g(\omega = 0)} = \frac{1}{200\,\mu\text{seg}} = 5000 \frac{\text{rad}}{s}$$
De este modo, todos los componentes se desnormalizan de la siguiente forma

Propongo como norma de impedancia $\Omega_Z = R = 1\,[\text{k}\Omega]$ para el circuito de primer orden y $\Omega_Z = R_1 = 1\,[\text{k}\Omega]$ para el circuito de segundo orden.
\begin{align}
    R &= R'\,\Omega_Z = 0.4306\times1\text{k}\Omega = 430.6\,\Omega\\\\
    C &= \frac{C'}{\Omega_Z\,\Omega_\omega} = \frac{1}{1000 \Omega \times 5000 \frac{\text{rad}}{s}} = 200\,\text{nF}\\\\
    R_1 &= R_1'\,\Omega_Z = 1 \text{k}\Omega \\\\
    R_2 &= R_2'\,\Omega_Z = 1 \text{k}\Omega \\\\
    C_1 &= \frac{C_1'}{\Omega_Z\,\Omega_\omega} = \frac{0.2847}{1000 \Omega \times 5000 \frac{\text{rad}}{s}} = 56.94\,\text{nF}\\\\
    C_2 &= \frac{C_2'}{\Omega_Z\,\Omega_\omega} = \frac{0.5438}{1000 \Omega \times 5000 \frac{\text{rad}}{s}} = 108.76\,\text{nF}\\\\
\end{align}

A continuación se realiza la simulación simbólica para comprobar que los cálculos realizados analíticamente anteriormente son correctos.

```python
import scipy.signal as sig
import numpy as np
from pytc2.sistemas_lineales import tf2sos_analog, pretty_print_SOS

# Definimos la transferencia de Bessel Orden 3 (normalizada)
order = 3
z, p, k = sig.besselap(order, norm='delay')
num, den = sig.zpk2tf(z, p, k)

# Ajuste de ganancia para que |T(0)| = 1
dc_gain = np.polyval(num, 0) / np.polyval(den, 0)
num_scaled = num / dc_gain

print("--- Transferencia Bessel Orden 3 ---")
print(f"Numerador: {num_scaled}")
print(f"Denominador: {den}")

# Factorización en Secciones de Segundo Orden (SOS)
sos = tf2sos_analog(num_scaled, den)

print("\n--- Factorización (SOS) para síntesis ---")
pretty_print_SOS(sos)
```
